```python
import gymnasium as gym 
```

```python
# Dummy Run
env = gym.make("LunarLander-v3", continuous=True, gravity=-10.0, 
                enable_wind=True, wind_power=15.0, turbulence_power=1.5,
                render_mode="human")

env.reset()

repeat = 3

print(env.action_space)

for episode in range(repeat):
    observation, info = env.reset()
    done = False
    total_reward = 0.0

    while not done:
        # Take a random action (later RL agent take action)
        action = env.action_space.sample()

        # Step forward in the environment
        observation, reward, terminated, truncated, info = env.step(action)
        total_reward += reward

        # Check if episode ended
        done = terminated or truncated
    
    print(f"Episode ended, total reward = {total_reward}")

env.close()
```
